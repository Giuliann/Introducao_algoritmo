## Introdução Curso de Algoritmo

### O que é um Algoritmo?

#### - Um algoritmo é uma sequência de passos lógicos a serem seguidos para solucionar um problema.

### Exemplo: 

    Algoritmo para atravessar a rua

        Olhar para a direita
        Olhar para a esquerda
        Se estiver vindo carro
            Não Atravesse
        senão
            Atravesse
        Fim-se
    Fim do Algoritmo

### Em outras palavras, um algoritmo é  em sua essencialmente uma receita, ou um conjunto estruturado de instruções a serem seguidas.

## Passos para criar um *Algoritmo Computacional*

O desenvolvimento de um algoritmo computacional começa com a criação de uma lógica de programação; depois, é necessário escrever essa lógica em uma linguagem de programação qualquer, como Python, Java, C, JavaScript etc. Utilizando uma das linguagens citadas, é possível criar um sistema completo, seja um aplicativo, uma ferramenta ou um site. Existem diversos métodos/técnicas de desenvolvimento; porém, diferentemente do curso em vídeo, que utiliza Portugol, irei utilizar a linguagem C.

## Comandos de Saida:

Em C, para escrever algo no terminal, utiliza-se a função `printf("Qualquer Coisa");`
O texto entre aspas duplas é uma string, ou cadeia de caracteres.

Exemplo em C:

    #include <stdio.h>
    #include <stdlib.h>

    int main(){
        printf("Ola, Mundo!");

        return 0;
    }

## O que são variáveis?

Variáveis são espaços de memória no computador utilizados para armazenar e consultar dados enquanto o programa funciona (está rodando ou sendo utilizado).
Para criar uma variável em C, é necessário informar o tipo da variável e o seu identificador.

Os identificadores, como o próprio nome deixa explícito, servem para identificar a variável e devem ser atribuídos a ela para que possa ser acessada na memória; porém, é necessário seguir algumas regras. Em C, as regras para nomear uma variável são:

* Começar obrigatoriamente com uma letra (maiúscula ou minúscula) ou com o caractere sublinhado (_).
* A linguagem C é case sensitive. Isso significa que variáveis como idade, Idade e IDADE são tratadas como três nomes completamente diferentes.
* Não são permitidos espaços em branco, acentos, símbolos ou caracteres especiais.
* Não se pode utilizar palavras-chave que pertencem à estrutura da linguagem C (como int, float, return, void, if e for).

Exemplos:

    salario | válido | Começa com letra e usa apenas caracteres permitidos.
    _taxa | válido | É permitido iniciar com sublinhado (_).
    nota1 | válido | Dígitos são permitidos após o primeiro caractere.

    2valor | inválido | Começa com um número (dígito).
    media final | inválido | Contém espaço em branco.
    peso-do-corpo | inválido | Usa hífen (-), que é um caractere especial proibido.

Os tipos primitivos em C são:

    * char | Caracteres individuais ou inteiros pequenos.
    * int | Números inteiros (positivos ou negativos).
    * float | Números reais de precisão simples (ponto flutuante).
    * double | Números reais de precisão dupla.
    * void | Tipo vazio.

Conhecendo as regras de nomeação de uma variável e os tipos primitivos, agora é possível declarar uma variável.

Exemplo de declaração de uma variável:

    #include <stdio.h>
    #include <stdlib.h>

    int main(){

        char mensagem[] = "Ola, Mundo!";

        printf("%s", mensagem);

        return 0;
    }

Note que foi utilizado o caractere "=", que serve para atribuir valores, ou seja, armazenar algo diretamente na variável.
